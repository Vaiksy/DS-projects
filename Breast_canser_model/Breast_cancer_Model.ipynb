```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from scipy.stats import zscore
```

```python
NNh = KNeighborsClassifier(n_neighbors= 5, weights= 'distance')
```

```python
bc_df = pd.read_csv("Ml ateek sir/00000050-wisc_bc_data-1 (1).csv")
```

```python
bc_df.shape
```

```python
bc_df.dtypes
```

```python
bc_df['diagnosis'] = bc_df.diagnosis.astype('category')
bc_df.dtypes
```

```python
bc_df.describe().transpose()
```

```python
bc_df.groupby(["diagnosis"]).count()
```

```python
bc_df = bc_df.drop(labels= "id", axis= 1)
```

```python
bc_df.shape
```

```python
bc_feature_df = bc_df.drop(labels= "diagnosis", axis= 1)
bc_feature_df.head()
```

```python
bc_feature_df_z = bc_feature_df.apply(zscore)
bc_feature_df_z.describe()
```

```python
bc_labels = bc_df["diagnosis"]
```

```python
X = np.array(bc_feature_df_z)
X.shape
```

```python
y = np.array(bc_labels)
y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.30, random_state=1)
```

```python
NNh.fit(X_train, y_train)
```

```python
predicted_labels = NNh.predict(X_test)
NNh.score(X_test, y_test)
```

```python
from sklearn import metrics
print(metrics.confusion_matrix(y_test, predicted_labels))
```

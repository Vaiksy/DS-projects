```python
df["waste_last_month"] = df["total_waste"].shift(1)
df["waste_3_months_ago"] = df["total_waste"].shift(3)
df["waste_6_months_ago"] = df["total_waste"].shift(6)
df["waste_last_year"] = df["total_waste"].shift(12)
```

```python
df = df.dropna()
```

```python
print(df.columns)
```

```python
df = df.drop(columns=["month"])
```

# Feature engineering 

```python
features = ["waste_last_month",
           "waste_3_months_ago",
            "waste_6_months_ago",
            "waste_last_year"
           ]

X = df[features]
y = df["total_waste"]
```

```python
train_size = int(len(df) * 0.8)

X_train = X[:train_size]
X_test = X[train_size:]

y_train = y[:train_size]
y_test = y[train_size:]
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_test, predictions)

rmse = np.sqrt(mean_squared_error(y_test, predictions))

print("MAE:", mae)
print("RMSE:", rmse)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))

plt.plot(y_test.index, y_test, label="Actual")
plt.plot(y_test.index, predictions, label="predictions")

plt.legend()
plt.title("Forecat VS Actual waste")
plt.show()
```

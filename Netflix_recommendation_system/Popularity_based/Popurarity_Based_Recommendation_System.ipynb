# Popurarity Based Recommendation System

```python
import numpy as np
import pandas as pd
```

```python
movie_names = pd.read_csv("movies.csv")
movie_names.head()
```

```python
rating_data = pd.read_csv("ratings.csv")
rating_data.head()
```

```python
movie_data = pd.merge(rating_data, movie_names, on='movieId')
```

```python
movie_data.head()
```

# Criteria For Popularity Based Recommendation Systems
The criteria is based on:
1. Movies with the highest rating
2. Number of views

# Average ratings of movies

```python
movie_data.groupby('title')['rating'].mean().head()
```

# Ratings count

```python
movie_data.groupby('title')['rating'].mean().sort_values(ascending=False).head(30)
```

# Rating Count

```python
movie_data.groupby('title')['rating'].count().sort_values(ascending=False).head(30)
```

# Lets add the average rating count to our table 

```python
movie_data[['title', 'rating']]
```

# Calculate the average ratings

```python
ratings_mean_count = pd.DataFrame(movie_data.groupby('title')['rating'].mean())
ratings_mean_count
```

# Calculate the Count of the ratings and add to the AVERAGE RATINGS above 

```python
ratings_mean_count['rating_counts'] = pd.DataFrame(movie_data.groupby('title')['rating'].count())
```

```python
ratings_mean_count
```

```python
ratings_mean_count['rating'] = round(ratings_mean_count['rating'], 1)
```

```python
ratings_mean_count.head()
```

```python
ratings_mean_count = ratings_mean_count[(ratings_mean_count['rating'] >4) & (ratings_mean_count['rating_counts'] >100)]
ratings_mean_count
```

```python
ratings_mean_count = ratings_mean_count.sort_values(by= 'rating', ascending=False).head(10)
ratings_mean_count
```
